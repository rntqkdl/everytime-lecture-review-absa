입력은 `segments_for_sentiment_colab.csv` 또는 별도 review-level 데이터다. segment-level 감정 파인튜닝은 수동 라벨 또는 매우 신뢰도 높은 pseudo-label이 있을 때만 수행한다.

```python
from pathlib import Path
import pandas as pd

INPUT_PATH = Path('/content/segments_for_sentiment_colab.csv')
OUTPUT_DIR = Path('/content/domain_finetuning_outputs')
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
df = pd.read_csv(INPUT_PATH, encoding='utf-8-sig')
df[['review_id','review_rating','rating_satisfaction_signal','rating_group']].head()
```

```python
# 안전 기본값: 실제 학습은 수행하지 않고 review-level 만족도 분류 실험 템플릿만 남긴다.
review_level = df.drop_duplicates('review_id').copy()
review_level[['review_id','review_rating','rating_satisfaction_signal','rating_group']].to_csv(
    OUTPUT_DIR / 'domain_finetuned_review_satisfaction_predictions.csv', index=False, encoding='utf-8-sig'
)
pd.DataFrame([{'status':'template_only','message':'Do not use review_rating as segment sentiment label.'}]).to_csv(
    OUTPUT_DIR / 'domain_finetuning_run_log.csv', index=False, encoding='utf-8-sig'
)
pd.DataFrame().to_csv(OUTPUT_DIR / 'domain_finetuned_model_metrics.csv', index=False, encoding='utf-8-sig')
print('template outputs saved')
```
